```python
import pandas as pd
df=pd.read_csv('luxury_cosmetics_popups.csv')
df
```

```python
df.info()
```

```python
df['start_date']=pd.to_datetime(df['start_date'])
```

```python
df['end_date']=pd.to_datetime(df['end_date'])
```

```python
df['revenue']=df['price_usd']*df['units_sold']
```

```python
df['city'].fillna('Unkown', inplace=True)
```

```python
df['end_date'].fillna(df['start_date'], inplace=True)
```

```python
df.info()
```

```python
df[['brand', 'product_name', 'price_usd', 'units_sold', 'revenue']].head(10)
```

```python
df.describe()
```

```python
df.groupby('brand')['price_usd'].sum().sort_values()
```

```python
brand_revenue=df.groupby('brand')['revenue'].sum().reset_index()
brand_revenue=brand_revenue.sort_values(by='revenue', ascending=False)
display(brand_revenue)
```

```python
top_brands=brand_revenue.sort_values(by='revenue', ascending=False).head(10)
top_brands
```

```python
city_revenue=df.groupby('city')['revenue'].sum().reset_index()
city_revenue=city_revenue.sort_values(by='revenue', ascending=False)

display(city_revenue)
```

```python
location_analysis=df.groupby('location_type').agg(
    total_revenue=('revenue', 'sum'),
    avg_revenue_per_revenue=('revenue', 'mean'),
    total_events=('event_id', 'count')
).reset_index().sort_values(by='total_revenue', ascending=False)

location_analysis
```

```python
top_products=df.groupby(['brand','product_name']).agg(
    total_units_sold=('units_sold', 'sum'),
    total_revenue=('revenue', 'sum')
).reset_index().sort_values(by='total_revenue', ascending=False)

top_products.head(20)
```

```python
correlation=df[['avg_daily_footfall', 'revenue', 'units_sold', 'price_usd']].corr()

correlation
```

```python
df['start_date']=pd.to_datetime(df['start_date'])
df['year_month']=df['start_date'].dt.to_period('M')

monthly_trend=df.groupby('year_month').agg(
    total_revenue=('revenue', 'sum'),
    total_events=('event_id', 'count'),
    avg_daily_footfall=('avg_daily_footfall', 'mean')
).reset_index()

monthly_trend['year_month']=monthly_trend['year_month'].astype(str)
monthly_trend
```

```python
event_analysis=df.groupby('event_type').agg(
    total_revenue=('revenue','sum'),
    avg_sell_through_pct=('sell_through_pct','mean'),
    total_units_sold=('units_sold','sum')
).reset_index().sort_values(by='total_revenue', ascending=False)

event_analysis
```

```python
brand_abc=df.groupby('brand')['revenue'].sum().reset_index()
brand_abc=brand_abc.sort_values(by='revenue', ascending=False)

brand_abc['cumulative_revenue']=brand_abc['revenue'].cumsum()
total_rev=brand_abc['revenue'].sum()
brand_abc['cum_percent']=(brand_abc['cumulative_revenue']/total_rev)*100

def assign_abc(percent):
    if percent <=80:
        return 'A (Top 80 %)'
    elif percent <=95:
        return 'B (Keyingi 15 %)'
    else:
        return 'C (Qolgan 5%)'

brand_abc['category']=brand_abc['cum_percent'].apply(assign_abc)
brand_abc
    
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(12,6))
sns.barplot(
    x='brand', y='revenue', data=top_brands, palette='Set2'
)
plt.title(
    'Top 10 Luxury Cosmetics Brand by Revenue',
    fontsize=14,
    fontweight='bold',
)
plt.xlabel('brand (Brend)', fontsize=12 )
plt.ylabel('revenue', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()
```
